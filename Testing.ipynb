```python
def factorial(n):
    if (n==0 or n==1):
        return 1
    else:
        return n * factorial(n-1)
print(factorial(5))
```

```python
#Armstrong Number
def IsArmstrongNumber(n):
    strNum=str(n)
    tot=0
    count=len(strNum)
    for i in strNum:
        tot+=int(i)**count
    return n==tot

print(IsArmstrongNumber(153))
```

```python
def find_most_frequent_manual(nums):
    if not nums:
        return None

    frequency_map = {}
    max_count = 0
    most_frequent_num = nums[0]

    # Populate the frequency map and track the maximum on the fly
    for num in nums:
        frequency_map[num] = frequency_map.get(num, 0) + 1
        print(frequency_map[num])
        if frequency_map[num] > max_count:
            max_count = frequency_map[num]
            most_frequent_num = num

    return most_frequent_num

# Example usage:
numbers = [5, 8, 5, 2, 5, 8, 8, 8, 1]
print(f"The most frequent number is: {find_most_frequent_manual(numbers)}")
# Output: 8
```

```python
import string

def is_pangram(sentence):
    # Convert the sentence to lowercase and turn it into a set
    sentence_set = sentence.lower()
    
    # Create a set of all 26 lowercase English alphabet letters
    alphabet_set = set(string.ascii_lowercase)
    
    # Check if the alphabet set is a subset of the sentence set
    return alphabet_set.issubset(sentence_set)

# Example usage:
phrase_1 = "The quick brown fox jumps over the lazy dog!"
phrase_2 = "Hello World"

print(f"Is phrase 1 a pangram? {is_pangram(phrase_1)}")  # Output: True
print(f"Is phrase 2 a pangram? {is_pangram(phrase_2)}")  # Output: False
```
