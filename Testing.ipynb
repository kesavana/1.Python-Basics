```python
def factorial(n):
    if (n==0 or n==1):
        return 1
    else:
        return n * factorial(n-1)
print(factorial(5))
```

```python
#Armstrong Number
def IsArmstrongNumber(n):
    strNum=str(n)
    tot=0
    count=len(strNum)
    for i in strNum:
        tot+=int(i)**count
    return n==tot

print(IsArmstrongNumber(153))
```
