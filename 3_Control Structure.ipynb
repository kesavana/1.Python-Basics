```python
age=int(input("Enter your age: "))
if(age<18):
    print("You're a childeren")
elif(age<35):
    print('You\'re a Adult')
elif(age<59):
    print("You're a Citizen")
else:
    print("You're a Senior Citizen")
```

```python
# We can use for loop for repitive task
lists =[12,34,55,45,56,78]
for age in lists:
    if(age<18):
        print("You're a childeren")
    elif(age<35):
        print('You\'re a Adult')
    elif(age<59):
        print("You're a Citizen")
    else:
        print("You're a Senior Citizen")
    
```

```python
# Get a number from the user to find whether Odd or Even number
# use format string to insert value of the variable in the string literal

number = int(input("Enter the number: "))
if(number%2==0):
    print(f"The given number {number} is even ")
else:
    print(f"The given number {number} is odd")
```

```python
# To find odd or even number using Ternary Operators
num=int(input("Enter the number: "))
result = "Odd" if (num%2==1) else "Even"
print(f"{num} is {result} number")
```

```python
# range(start, stop,step)
# To find odd number between 1 to 10 using range
for num in range(1,10):
    if(num%2==1):
        print(num)
```

```python
# Print multiplication table 
# Get table input from user
table=int(input("Enter table number: "))
for num in range(1,11):
    print(f"{num} X {table} =", (num * table))
```
