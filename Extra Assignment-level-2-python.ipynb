```python
#Print Odd numbers in the list
list1=[20,10,16,19,25,1,276,188]
for num in list1:
    if(num%2==1):
        print(f"{num} is Odd")
```

```python
#Print Even Odd numbers in the list
list1=[20,10,16,19,25,1,276,188]
for num in list1:
    if(num%2==0):
        print(f"{num} is Even")
```
