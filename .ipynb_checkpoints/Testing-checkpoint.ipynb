```python
def factorial(n):
    if (n==0 or n==1):
        return 1
    else:
        return n * factorial(n-1)
print(factorial(5))
```

```python
#Armstrong Number
def IsArmstrongNumber(n):
    strNum=str(n)
    tot=0
    count=len(strNum)
    for i in strNum:
        tot+=int(i)**count
    return n==tot

print(IsArmstrongNumber(153))
```

```python
str1=input("Enter the string values: ")
word=str1.split()
temp=0
long=""
for i in word:
    if(temp>int(len(i))):
        temp=temp
        long=i
    else:
        temp=len(i)
        long=i
print(f"The longest word is {long} and it's length is {temp}")

print(max(word,key=len)) # Shortest way to find the longest word in the given sentences
    
```
