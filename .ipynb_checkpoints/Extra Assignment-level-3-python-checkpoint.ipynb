```python
# Check the password using if and else
pwd=input("Enter your password: ")
if(pwd=="Hope@123"):
    print("Your password is correct")
else:
    print("Your password is incorrect")
```

```python
#Category the people by their age like children, adult, citizen and senior citizen
age=int(input("Enter your age :"))
if(age<18):
    print(f"Your age is {age}. You're come under Children category")
elif(age<35):
    print(f"Your age is {age}. You're come under Adult category")
elif(age<55):
    print(f"Your age is {age}. You're come under Citizen category")
else:
    print(f"Your age is {age}. You're come under Senior Citizen category")
```

```python
# Find whether given number is Positive or Negative
num=int(input("Enter the number: "))
if(num>0):
    print(f"The given number {num} is Positive")
elif(num==0):
    print(f"The given number {num} is Neutral")
else:
    print(f"The given number {num} is Negative")
    
```

```python
#Check whether given number is divisible by 5
num=int(input("Enter the number: "))
if(num%5==0):
    print(f"The given number {num} is divisible by 5")
else:
    print(f"The given number {num} is not divisible by 5")
```
