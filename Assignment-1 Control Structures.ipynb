```python
# Classify based on Body Mass Index (BMI) value
# Get BMI value from the user
BMI=float(input("Enter your BMI value: "))
if(BMI<=15):
    print(f" BMI ({BMI}) = Very serverly Underweight")
elif(BMI<=16):
    print(f" BMI ({BMI}) = Serverly Underweight")
elif(BMI<=18.5):
    print(f" BMI ({BMI}) = Underweight")
elif(BMI<=25):   
    print(f" BMI ({BMI}) = Normal (Healthy Weight)")
elif(BMI<=30):
    print(f" BMI ({BMI}) = OVerweight")
elif(BMI<=35):
    print(f" BMI ({BMI}) = Obese Class -I (Moderately Obese)")
elif(BMI<=40):
    print(f" BMI ({BMI}) = Obese Class -II (Severly Obese)")
else:
    print(f" BMI ({BMI}) = Obese Class -III (Very Severly Obese)")
    
```
