```python
#3.Write a Python function to check if a given string is a palindrome.
#Expected Output (for input "madam"):
#madam is a palindrome
originalText=input("Enter your string: ")
reverseText=""
for i in range(len(originalText)):
    reverseText=originalText[i]+reverseText
if(originalText.lower() == reverseText.lower()):
    print(f"{reverseText} is a palindrome")
else:
    print(f"{reverseText} is not a palindrome")
```

```python
#4.Write a Python program to count the number of vowels in a given string. 
#Expected Output (for input "education"):
#Number of vowels: 5
    
count=0
str1=input("Enter the string value : ")
for i in range(len(str1)):
    if(str1[i]=='a' or str1[i]=='e' or str1[i]=='i' or str1[i]=='o' or str1[i]=='u' or str1[i]=='A' or str1[i]=='E' or str1[i]=='I' or str1[i]=='O' or str1[i]=='U'):
        count+=1
print(f"The given string {str1} contains {count} vowels")
```
