```python
#5.What is the issue with this function?

def is_prime(n):
    for i in range(2, n):
        if n % i == 0:
            return False  #Return statement wrongly mentioned as True instead of False
    return True
is_prime(7)
```

```python
#8.What is the issue with this function?

def sum_numbers(numbers):
    total = 1
    for num in str(numbers): #cannot iterate integer number directly, so first convert the number to string
        total += int(num) # convert string literal value into int
    return total

print(sum_numbers(234))
```

```python
#9.What is the issue with this recursive function?

def factorial(n):
    if n <=1:   
        return 1
    return n * factorial(n-1) # Recursiion Error, we need to define the stop condition when to stop the execution

print(factorial(5))
```
