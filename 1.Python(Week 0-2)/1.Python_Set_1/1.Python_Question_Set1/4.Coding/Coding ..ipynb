```python
#4.Write a Python function to find the factorial of 5. Expected Output:
#120 5*4*3*2*1

def factorial():
    num=int(input("Enter the factorial number: "))
    fact=1
    if(num==0):
        print("1")
    elif(num<1):
        print("Value must be positive number")
    else:
        for i in range(1,num+1):
            fact*=(i)
        print(fact)

factorial()
```

```python
#5.Write a Python program to check if a number is prime. Expected Output (for input 7):
#7 is a prime number

def primenumber():
    num=int(input("Enter the number to check prime or not: "))
    if(num<=1):
        return False
    elif(num==2):
        return True
    elif(num%2==0):
        return False
    else:
        limit=int(num**0.5)
        #print(limit)
        for i in range(3,limit+1,2):
            if(num%i==0):
                return False
        return True
                
status=primenumber()
if(status==True):
    print("The given number is prime")
else:
    print("The given number is not prime")
```
