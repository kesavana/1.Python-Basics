```python
#2.Swap Two Variables Without Using a Temporary Variable
#Question: Write a Python program to swap two variables a = 5 and b = 10 without using a temporary variable.
a=5
b=10
b=a+b
a=b-a
b=b-a
print(a)
print(b)

# Other way to swap. Python allow multiple assign variable at the same time
x=5
y=10
x,y =y,x
print(f" x = {x}")
print(f" y = {y}")
```

```python
 #3.Find Even and Odd Numbers in a List
#Question: Write a Python program to separate even and odd numbers from the list [12, 17, 19, 24, 35, 40, 56].
original=[12, 17, 19, 24, 35, 40, 56]
odd=[]
even=[]
for i in original:
    if(i%2==0):
        even.append(i)
    else:
        odd.append(i)

print(odd)
print(even)
# Other approch using list comprehension
list1=[12, 17, 19, 24, 35, 40, 56]
a=[i for i in list1 if i%2==0]
b=[i for i in list1 if i%2==1]
print(a)
print(b)
```

```python
#4. Reverse a Number
#Question: Write a Python program to reverse the number 12345.
num=12345
str1=str(num)
rev=""
for i in str1:
    rev=i+rev
print(int(rev))
print(int(str(num)[::-1]))#shortest way to reverse number
```

```python
#5. Find the Second Largest Number in a List
#Question: Write a Python program to find the second largest number in the list [12, 45, 98, 34, 23].
original=[12, 45, 98, 34, 23]
original.sort()
print(original[-2])

largest=float('-inf') # initialize negative very low value 
second_largest=float('-inf')
for num in original:
    if num>largest:
        second_largest=largest
        largest=num
    elif num>second_largest and num!= largest:
        second_largest=num
print(f"The second largest number in the list is {second_largest}")
```
