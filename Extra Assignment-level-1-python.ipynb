```python
# Find Two numbers of Addition, Subtraction, Multiplication, Division, Modulo, Floor Division, Power 
import math
a=float(input("Enter A value : "))
b=float(input("Enter B value : "))
print(f" Value of {math.floor(a)} + {math.floor(b)} is {math.floor(a+b)}")
print(f" Value of {math.floor(a)} - {math.floor(b)} is {math.floor(a-b)}")
print(f" Value of {math.floor(a)} * {math.floor(b)} is {math.floor(a*b)}")
print(f" Value of {(a)} / {(b)} is {(a/b)}")
print(f" Value of {math.floor(a)} / {math.floor(b)} is {math.floor(a/b)}")
print(f" Value of {math.floor(a)} % {math.floor(b)} is {math.floor(a%b)}")
print(f" Value of {math.floor(a)} + {math.floor(b)} is {math.floor((a**b))}")
```
